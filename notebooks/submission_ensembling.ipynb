```python
import pandas as pd

print("Starting the Grandmaster Blend...")

# 1. Load the two submission CSV files
# (Make sure the file names match what you actually saved them as!)
my_model = pd.read_csv('v1_submission.csv')
chris_model = pd.read_csv('submission.csv')

# 2. Verify they have the exact same rows and Passenger IDs
assert len(my_model) == len(chris_model), "Error: Files have different lengths!"
assert (my_model['id'] == chris_model['id']).all(), "Error: IDs do not match!"

# 3. Create a copy for the final blended submission
final_ensemble = my_model.copy()

# 4. THE BLEND: Weighted Average
# Chris's model is stronger, so we give his vote 75% weight. 
# Your XGBoost model gets 25% weight to catch the edge cases.
final_ensemble['addicted_label'] = (my_model['addicted_label'] * 0.25) + (chris_model['addicted_label'] * 0.75)

# 5. Save the final masterpiece
final_ensemble.to_csv('final_ensemble_submission.csv', index=False)

print("🎉 Final Ensemble CSV created! Ready for Kaggle Submission.")
```
